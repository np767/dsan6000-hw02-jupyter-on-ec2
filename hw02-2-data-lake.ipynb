```python
import os
import boto3
```

```python
#define netid and bucket name
netid = "np767"
bucket_name = f"dsan6000-{netid.lower()}"
region = "us-east-1"

#initialize standard authenticated Boto s3 client
s3_client = boto3.client("s3", region_name=region)

#Create the bucket 
response = s3_client.create_bucket(Bucket=bucket_name)

print(f"Successfully created bucket: s3://{bucket_name} in region {region}")
```

```python
#iterate over local files and upload them to the S3 bucket
local_data_dir = "data"
s3_prefix = "wikipedia-hourly"

#get list of local parquet files
files_to_upload = [f for f in os.listdir(local_data_dir) if f.endswith(".parquet")]

for file_name in files_to_upload:
    local_file_path = os.path.join(local_data_dir, file_name)
    s3_key = f"{s3_prefix}/{file_name}"
    
    #upload the file to S3
    s3_client.upload_file(local_file_path, bucket_name, s3_key)
    print(f"Uploaded {local_file_path} to s3://{bucket_name}/{s3_key}")
    
print(f"Successfully uploaded {len(files_to_upload)} files to s3://{bucket_name}/{s3_prefix}/")
```

```python
#verify files in s3 bucket
response = s3_client.list_objects_v2(Bucket=bucket_name, Prefix=s3_prefix)
contents = response.get('Contents', [])

print(f"total uploaded files found in s3://{bucket_name}/{s3_prefix}/: {len(contents)}")
for obj in contents[:5]:
    print(f" - {obj['Key']} (size: {obj['Size']} bytes)")
```
