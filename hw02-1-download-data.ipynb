```python
import os
import boto3
from botocore import UNSIGNED
from botocore.config import Config
```

```python
#Create data folder if it does not exist
local_data_dir = "data"
os.makedirs(local_data_dir, exist_ok=True)
```

```python
#Initialize S3 Client for public access
s3_client = boto3.client('s3', config=Config(signature_version=UNSIGNED))

bucket_name = "dsan6000-wikipedia"
prefix = "hourly_parquet/"

#list objects in the source bucket prefix
response = s3_client.list_objects_v2(Bucket=bucket_name, Prefix=prefix)

#Download each file into the local data directory
for obj in response.get("Contents", []):
    key = obj["Key"]
    
    #Skip the folder prefix
    if key.endswith("/") or not key.endswith(".parquet"):
        continue
    
    file_name = os.path.basename(key)
    local_file_path = os.path.join(local_data_dir, file_name)
    
    print(f"Downloading {key} to {local_file_path}...")
    s3_client.download_file(bucket_name, key, local_file_path)
    
print("Download complete.")
```

```python
#Final check to make sure all data downloaded and lives inside data/
downloaded_files = os.listdir(local_data_dir)
print(f"Total files downloaded: {len(downloaded_files)}")
print(f"Sample files: {downloaded_files[:5]}")
```
